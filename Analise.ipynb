```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
paths = { 
    2022: [
        r"D:\Projeto_Supervisionado\taxi-chicago-analysis\data\raw\taxi_trips\year=2022\month=01\data.parquet",
        r"D:\Projeto_Supervisionado\taxi-chicago-analysis\data\raw\taxi_trips\year=2022\month=06\data.parquet",
        r"D:\Projeto_Supervisionado\taxi-chicago-analysis\data\raw\taxi_trips\year=2022\month=12\data.parquet",
    ],
    2023: [
        r"D:\Projeto_Supervisionado\taxi-chicago-analysis\data\raw\taxi_trips\year=2023\month=01\data.parquet",
        r"D:\Projeto_Supervisionado\taxi-chicago-analysis\data\raw\taxi_trips\year=2023\month=06\data.parquet",
        r"D:\Projeto_Supervisionado\taxi-chicago-analysis\data\raw\taxi_trips\year=2023\month=12\data.parquet",
    ],
    2024: [
        r"D:\Projeto_Supervisionado\taxi-chicago-analysis\data\raw\taxi_trips\year=2024\month=01\data.parquet",
        r"D:\Projeto_Supervisionado\taxi-chicago-analysis\data\raw\taxi_trips\year=2024\month=06\data.parquet",
        r"D:\Projeto_Supervisionado\taxi-chicago-analysis\data\raw\taxi_trips\year=2024\month=12\data.parquet",
    ],
    2025: [
        r"D:\Projeto_Supervisionado\taxi-chicago-analysis\data\raw\taxi_trips\year=2025\month=01\data.parquet",
        r"D:\Projeto_Supervisionado\taxi-chicago-analysis\data\raw\taxi_trips\year=2025\month=06\data.parquet",
        r"D:\Projeto_Supervisionado\taxi-chicago-analysis\data\raw\taxi_trips\year=2025\month=12\data.parquet",
    ]
}

dfs_por_ano = {}

for ano, arquivos in paths.items():
    dfs = [pd.read_parquet(p) for p in arquivos]
    dfs_por_ano[ano] = pd.concat(dfs, ignore_index=True)
```

```python
print('-' * 30, "SHAPE DOS DADOS", '-' * 30)
print("Ano 2022: ", dfs_por_ano[2022].shape)
print("Ano 2023: ", dfs_por_ano[2023].shape)
print("Ano 2024: ", dfs_por_ano[2024].shape)
print("Ano 2025: ", dfs_por_ano[2025].shape)
```

```python
pd.set_option('display.max_rows', None)
pd.set_option('display.max_columns', None)
print('-' * 30, "Dados Nulos", '-' * 30)
for ano, df in dfs_por_ano.items():
    nulos = df.isnull().sum()
    nulos = nulos[nulos > 0]

    perc = (nulos / len(df)) * 100

    print(f"\nAno {ano}")
    
    if nulos.empty:
        print("Sem valores nulos 🎉")
    else:
        print(pd.DataFrame({
            "nulos": nulos,
            "%": perc.round(2)
        }).sort_values("nulos", ascending=False))
```

```python
print('-' * 30, "Coordenadas Inválidas", '-' * 30)

for ano, df in dfs_por_ano.items():
    invalid = df[
        (df['pickup_centroid_latitude'] < 41.6) | (df['pickup_centroid_latitude'] > 42.1) |
        (df['pickup_centroid_longitude'] < -87.9) | (df['pickup_centroid_longitude'] > -87.5)
    ]
    
    print(f"Ano {ano}: {len(invalid)} registros inválidos")
```

```python
print('-' * 30, "Outliers", '-' * 30)


for ano, df in dfs_por_ano.items():
    df['trip_miles'] = pd.to_numeric(df['trip_miles'], errors='coerce')
    df['trip_seconds'] = pd.to_numeric(df['trip_seconds'], errors='coerce')
    df['fare'] = pd.to_numeric(df['fare'], errors='coerce')

    outliers = df[
        (df['trip_miles'] < 0) | (df['trip_miles'] >= 100) |
        (df['trip_seconds'] < 60) | (df['trip_seconds'] > 7200) |
        (df['fare'] < 2.5) | (df['fare'] > 500)
    ]

    print(f"Ano {ano}: {len(outliers)} outliers")
```

```python
print('-' * 30, "Duplicatas por trip_id", '-' * 30)

for ano, df in dfs_por_ano.items():
    dup_rows = df[df['trip_id'].duplicated(keep=False)]
    print(f"Ano {ano}: {len(dup_rows)} linhas duplicadas")
```

```python
print('-' * 30, "Inconsistência Temporal", '-' * 30)

for ano, df in dfs_por_ano.items():
    df['trip_start_timestamp'] = pd.to_datetime(df['trip_start_timestamp'])
    df['trip_end_timestamp'] = pd.to_datetime(df['trip_end_timestamp'])

    invalid_time = df[df['trip_end_timestamp'] < df['trip_start_timestamp']]
    
    print(f"Ano {ano}: {len(invalid_time)} inconsistências")
```

```python
print('-' * 30, "Resumo Percentual", '-' * 30)

for ano, df in dfs_por_ano.items():
    total = len(df)

    outliers = df[
        (df['trip_miles'] < 0) | (df['trip_miles'] >= 100) |
        (df['trip_seconds'] < 60) | (df['trip_seconds'] > 7200) |
        (df['fare'] < 2.5) | (df['fare'] > 500)
    ]

    perc = (len(outliers) / total) * 100

    print(f"Ano {ano}: {perc:.2f}% outliers")
```
